# GPT-4o

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import kendalltau

# CSV 파일 불러오기
df = pd.read_csv('results/ea_prompt_gpt-4o_fewshot3_results.csv')

# 필요한 열만 추출 (열 이름은 실제 파일에 따라 조정 필요)
cols = ['MQMScore', 'LLM_MQM_gemba_mqm', 'LLM_MQM_ea_prompt']
df = df[cols].dropna()  # 결측치 제거

# Kendall's Tau 상관계수 계산
tau_gemba, p_gemba = kendalltau(df['MQMScore'], df['LLM_MQM_gemba_mqm'])
tau_ea, p_ea = kendalltau(df['MQMScore'], df['LLM_MQM_ea_prompt'])

print(f"Kendall's Tau (Gemba): {tau_gemba:.3f}, p-value: {p_gemba:.3f}")
print(f"Kendall's Tau (EA): {tau_ea:.3f}, p-value: {p_ea:.3f}")

# 시각화
sns.pairplot(df)
plt.suptitle("Score Distribution & Correlation")
plt.show()
```
